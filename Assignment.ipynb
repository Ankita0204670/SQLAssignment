```python
import pandas as pd
import sqlite3

excel_file = 'SQL_Sales_Dataset_200_Rows.xlsx'
df = pd.read_excel(excel_file)

print("Data preview:")
display(df.head())

conn = sqlite3.connect(':memory:')
df.to_sql('sales', conn, index=False, if_exists='replace')

print("\n--- TOP 10 CUSTOMERS BY TOTAL SPEND ---")
query_top_customers = """
    SELECT
        customer_name,
        SUM(total_price) AS total_spend,
        COUNT(order_id) AS order_count
    FROM
        sales
    GROUP BY
        customer_name
    ORDER BY
        total_spend DESC
    LIMIT 10;
"""
top_customers_df = pd.read_sql_query(query_top_customers, conn)
display(top_customers_df)

print("\n--- OVERALL AVERAGE ORDER VALUE (AOV) ---")
query_aov = """
    SELECT
        AVG(total_price) AS overall_aov,
        SUM(total_price) AS total_revenue,
        COUNT(order_id) AS total_orders
    FROM
        sales;
"""
aov_df = pd.read_sql_query(query_aov, conn)
display(aov_df)

conn.close()
```
